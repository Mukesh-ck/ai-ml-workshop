```python
import numpy as np
import pandas as pd

data = {
    'student_name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve','Alice'],
    'hours_studied': [5,8,3,10,6,5],
    'previous_score': [65, 85, 55, 92, 75, 65],
    'attendance': ['Yes', 'No', 'Yes', 'NO', 'Yes', 'Yes'],
    'sleep_hours': [7, 5, 4, 7, 6, 7],
    'extracurricular': ['Yes', 'No', 'Yes', 'NO', 'Yes','Yes'],
    'parent_education': ['Bachelor', 'Master', 'High School', 'PhD', 'Bachelor','Bachelor']
}

df = pd.DataFrame(data)
print("Student Performance DataFrame:")
print(df)

#Basic DataFrame Information 
print("\nDataFrame Shape:", df.shape[0], "rows,", df.shape[1], "columns")
print("Column names:", df.columns.tolist())
print("Data types:\n", df.dtypes)
```

```python
# view data
print("First 5 rows:")
print(df.head())

# view last 2 rows 
print("last 3 rows:")
print(df.tail(2))

# Statistical summary
print("\nStatistical Summary:")
print(df.describe())

# Check for missing values
print("\nMissing values:")
print(df.isnull().sum())

# Check for nan
print("\nCheck nan values")
print(df.isna().sum())

# check for dulicate rows
print("\nDuplicate Rows: ")
print(df.duplicated().sum())
```
